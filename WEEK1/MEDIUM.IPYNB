```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

# 1. Create a 1-qubit circuit with 1 classical bit for measurement
qc = QuantumCircuit(1, 1)

# 2. Apply a Hadamard gate and measure
qc.h(0)
qc.measure(0, 0)

# Display the colourful circuit
display(qc.draw('mpl', style='clifford'))
```

```python
# 3. Execute 1024 shots on the ideal simulator
simulator = AerSimulator()
compiled_circuit = transpile(qc, simulator)
job = simulator.run(compiled_circuit, shots=1024)
counts = job.result().get_counts()

print("Measurement Counts:", counts)

# 4. Plot the histogram
plot_histogram(counts, title="1-Qubit Superposition", color='midnightblue')
```
