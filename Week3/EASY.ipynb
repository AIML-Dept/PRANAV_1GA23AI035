```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator

# ==============================================================
# WEEK 3 - EASY
# Implement the Pauli-X, Pauli-Y and Pauli-Z gates individually,
# each on its own qubit, and print the resulting statevectors.
# ==============================================================


def to_ket(sv):
    """Return a readable Dirac-notation string for a 1-qubit statevector."""
    a, b = np.round(sv.data, 4)
    return f"({a:.4f})|0> + ({b:.4f})|1>"


# Each gate gets its own single-qubit circuit (its own qubit)
gates = {
    "Pauli-X  (bit flip / quantum NOT)": lambda qc: qc.x(0),
    "Pauli-Y  (bit flip + phase flip)": lambda qc: qc.y(0),
    "Pauli-Z  (phase flip)": lambda qc: qc.z(0),
}

# The two computational basis states we will feed into each gate
inputs = {"|0>": Statevector([1, 0]), "|1>": Statevector([0, 1])}

circuits = {}

for name, apply_gate in gates.items():
    qc = QuantumCircuit(1)
    apply_gate(qc)
    circuits[name] = qc

    print("=" * 62)
    print(f"GATE: {name}")
    print("=" * 62)

    # --- Gate matrix -------------------------------------------------
    matrix = Operator(qc).data
    print("Gate matrix:")
    print(np.round(matrix, 3))

    # --- Action on |0> (the default statevector of a fresh circuit) ---
    default_state = Statevector.from_instruction(qc)
    print("\nResulting Statevector (circuit starts in |0>):")
    print(default_state)

    # --- Action on both basis states ---------------------------------
    print("\nAction on the basis states:")
    for label, in_state in inputs.items():
        out_state = in_state.evolve(qc)
        probs = np.abs(out_state.data) ** 2
        print(f"  {label} -> {to_ket(out_state)}"
              f"   | probabilities: P(0)={probs[0]:.2f}, P(1)={probs[1]:.2f}")
    print()

# ------------------------------------------------------------------
# Interpretation
# ------------------------------------------------------------------
print("=" * 62)
print("INTERPRETATION")
print("=" * 62)
print("X : swaps |0> and |1>  -> a rotation of 180 degrees about the X axis.")
print("Y : swaps |0> and |1> AND adds an i / -i phase -> 180 deg about Y.")
print("Z : leaves |0> alone, multiplies |1> by -1 -> 180 deg about Z.")
print("\nNote: Z acting on |0> looks like 'nothing happened' because the")
print("phase it adds only shows up on the |1> component. Measurement")
print("probabilities never change under Z alone - only the relative phase.")

# --- Sanity check: all three Paulis square to the identity ---------
print("\n--- Verification: P^2 = I for every Pauli gate ---")
I = np.eye(2)
all_ok = True
for name, qc in circuits.items():
    M = Operator(qc).data
    ok = np.allclose(M @ M, I)
    all_ok &= ok
    print(f"  {name.split()[0]:8s} squared == Identity ? {ok}")

print("\nSTATUS: SUCCESS - all Pauli gates are their own inverse."
      if all_ok else "\nSTATUS: FAILED")

# ------------------------------------------------------------------
# Circuit diagrams
# ------------------------------------------------------------------
print("\n--- Quantum Circuits ---")
for name, qc in circuits.items():
    print(name)
    display(qc.draw(output="mpl", style="iqp"))
```
