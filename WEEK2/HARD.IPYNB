```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# 1. Initialize a 2-qubit quantum circuit
qc = QuantumCircuit(2)

# 2. Apply Hadamard gate (H) to both qubits
# Applying H to both creates an equal superposition of all 4 basis states
qc.h(0)
qc.h(1)

# 3. Compute the resulting Statevector
state = Statevector.from_instruction(qc)

print("Resulting Statevector:")
print(state)
print("\nIndividual Amplitudes for basis states [|00>, |01>, |10>, |11>]:")
for idx, amp in enumerate(state.data):
    print(f"  State |{format(idx, '02b')}> : amplitude = {amp:.4f}, probability = {abs(amp)**2:.4f}")

# 4. Verify that the sum of absolute squared amplitudes equals 1 (Unity)
total_probability = np.sum(np.abs(state.data) ** 2)

print("\n--- Verification ---")
print(f"Sum of squared magnitudes (Total Probability): {total_probability:.6f}")

if np.isclose(total_probability, 1.0):
    print("STATUS: SUCCESS - The amplitudes sum to unity! Quantum state is valid and normalized.")
else:
    print("STATUS: FAILED - State is not normalized properly.")

# 5. Display the colorful circuit diagram inline in Jupyter Notebook
display(qc.draw(output='mpl', style='iqp'))
```
