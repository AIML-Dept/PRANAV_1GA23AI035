```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector

# 1. Create a 1-qubit circuit
qc = QuantumCircuit(1)

# 2. Apply Hadamard (H) gate followed by Phase (S) gate
qc.h(0)
qc.s(0)

# 3. Compute and print the resulting Statevector
state = Statevector.from_instruction(qc)
print("Resulting Statevector:")
print(state)

# 4. Display the colorful circuit diagram inline (No PNG saving)
display(qc.draw(output='mpl', style='iqp'))

# 5. Display the Bloch Sphere inline
plot_bloch_multivector(state)
```
